```python


def calculate_total(amount, Trate = 0.05):
    return amount * (1 + Trate)


print(f"Total: {calculate_total(100):.2f}")
print(f"Total: {calculate_total(200, 0.12):.2f}")
```

```python

price = float(input("Enter the original price: "))


def apply_discount(price, discount_percent):

    return price * (1 - discount_percent / 100)

res = apply_discount(price, 10) 
print(f"price + discount: {res:.2f}")

```

```python
def basic_stats(numbers):
    count = len(numbers)
    total = sum(numbers)
    mean = total / count if count > 0 else None
    return count, total, mean

print(basic_stats([10, 20, 30]))
```

```python
def apply_tax(amounts, tax_rate=0.1):
    return [round(amt * (1 + tax_rate), 2) for amt in amounts]

print(apply_tax([100, 200, 350], 0.05))
```

```python
def process_score(score, max_score=100):
    if score < 0 or score > max_score:
        return None
    return round((score / max_score) * 100, 1)

print(process_score(85))
print(process_score(150))
```
