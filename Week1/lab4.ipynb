```python
import pandas as pd

df = pd.read_csv("E:\\Data_Engineering_Labs\\CSV2\\nba.csv")
```

```python
print(df.head(5))
print(df.shape)
print(df.columns)
```

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
print(df.isna().sum())
df = df.dropna(subset=["Salary"])
df["College"] = df["College"].fillna("Unknown")
```

```python

celtics_players = df[df["Team"] == "Boston Celtics"]
print(celtics_players)
```

```python
players_over_30 = df[df["Age"] > 30]
print(players_over_30)
```

```python
pg_players = df[df["Position"] == "PG"]
print(pg_players)
```

```python
p_salary = df[df["Salary"] > 5000000]
print(p_salary)
```

```python
df_sorted= df.sort_values   (by="Salary", ascending=False)
print(df_sorted)
```

```python
df_sorted = df.sort_values(by="Salary", ascending=True)
print(df_sorted)
```
