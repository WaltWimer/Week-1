```python
sales = {"Asha": 1200.50, "Rohan": 250.00, "Tina": 799.99}
total = 0.0
for name, amount in sales.items():
    total += amount
print(f"Total sales: {total}")
```

```python
totals = {"Asha": 1200.50, "Rajan": 3200.00, "Leela": 2450.75}
max_name = ""
max_amount = 0.0
for name, amount in totals.items():
    if amount > max_amount:
        max_amount = amount
        max_name = name
print(f"Top customer: {max_name} with {max_amount}")
```

```python
balance = 500.0
transactions = [100.0, -50.0, 0.0, -600.0, 200.0]
i = 0
while i < len(transactions):
    amount = transactions[i]
    if amount == 0:
        i += 1
        continue
    balance += amount
    if balance < 0:
        print("Processing stopped early due to negative balance.")
        break
    i += 1
print(f"Final balance: {balance}")
```
